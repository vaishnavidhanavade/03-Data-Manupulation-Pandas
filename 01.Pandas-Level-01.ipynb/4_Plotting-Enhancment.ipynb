```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_excel(r"D:\ethans\Ethans\Swarn_Airline (3).xlsx")
df.head(2)
```

```python
order = df.groupby('Airline')['Price'].sum().sort_values().index
order
```

```python
sns.set_theme(style="dark",palette='pastel')
plt.figure(figsize=(10,5))

Airline_Sort = df.groupby('Airline')['Price'].sum(),sort_values().index
ax=sns.barplot(x=df['Airline'],y=df['Price'],errorbar=None,estimator='mean',legend=False,order=Airline_Sort,alpha=9,gap=.1)
ax.bar_label(ax.containers[0],rotation=80)
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
sns.set_theme(style="darkgrid", palette="Set1")
plt.figure(figsize=(10,5))
Airline_Sort = df.groupby('Airline')['Price'].sum().sort_values().index

ax = sns.countplot(x=df['Airline'],legend=False, order=Airline_Sort, alpha=.9, gap=.1)
ax.bar_label(ax.containers[0], rotation=90)

plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
sns.set_theme(style="whitegrid", palette="RdBu_r")
plt.figure(figsize=(10,5))

AirLine_sort = df.groupby('Airline')['Price'].sum().sort_values().index

ax=sns.countplot(x=df['Airline'], order=AirLine_sort, alpha=0.8, gap=0.3, hue=df['Source'], legend=True, palette='viridis')
ax.bar_label(ax.containers[0], rotation=90)

plt.xticks(rotation=60)
plt.tight_layout()
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
plt.figure(figsize=(10,5))

Airline_Sort = df['Airline'].value_counts().index

ax = sns.countplot( data=df,x='Airline',order=Airline_Sort,alpha=0.9,gap=0.1)

plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```
