```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.figure(figsize=(12,6))

sns.countplot(data=df, x='Airline')

plt.xticks(rotation=45)
plt.xlabel("Airline")
plt.ylabel("Count")
plt.title("Airline Count Plot")

plt.show()
```
