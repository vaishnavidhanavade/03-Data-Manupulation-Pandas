```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r"D:\ethans\Ethans\Airline1.csv")
```

```python
plt.figure(figsize=(12,6))

plt.plot(df['Airline'], df['Price'], marker='o')

plt.xticks(rotation=45)
plt.xlabel("Airline")
plt.ylabel("Price")
plt.title("Airline vs Price")

plt.grid()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.figure(figsize=(12,6))

sns.countplot(data=df, x='Airline')

plt.xticks(rotation=45)
plt.xlabel("Airline")
plt.ylabel("Count")
plt.title("Airline Count Plot")

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.figure(figsize=(12,6))

sns.violinplot(data=df, x='Airline', y='Price')

plt.xticks(rotation=45)
plt.xlabel("Airline")
plt.ylabel("Price")
plt.title("Airline vs Price Violin Plot")

plt.show()
```
